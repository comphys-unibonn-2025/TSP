```python
# points on a circle
N = 8
R = 1
phi = np.array([2 * np.pi * i / N + np.pi / 2 for i in range(0,N)])

nodes_circle = polar_to_cartesian(R, phi)
```

```python
plot_tsp_route(nodes_circle, np.zeros(N, dtype=int))
```

```python
optimal_circle_route = simmulated_annealing_tsp(nodes_circle, plot_cost=True)
```

```python
print("Optimal route for points on a circle:", optimal_circle_route)
plot_tsp_route(nodes_circle, optimal_circle_route)
```
