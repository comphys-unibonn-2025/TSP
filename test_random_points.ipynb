```python
from tsp_backend import simmulated_annealing_tsp, plot_tsp_route, constant_traffic_factor_func
import numpy as np

N = 100
x = np.random.rand(N)
y = np.random.rand(N)
nodes_random = np.column_stack([x, y])

def traffic_factor_no_traffic(cost : float) -> float:
    return constant_traffic_factor_func(1.0, 0.0, cost)

optimal_route_no_traffic = simmulated_annealing_tsp(nodes_random, traffic_factor_no_traffic, plot_cost=True)
```

```python
plot_tsp_route(nodes_random, optimal_route_no_traffic)
```

```python
traffic_jam_area = np.array([[0.1, 0.1], [0.9, 0.1], [0.9, 0.9], [0.1, 0.9]]) # rectangle
traffic_factor = 100.0
traffic_start_time = 10.0

def traffic_factor_with_traffic(cost : float) -> float:
    return constant_traffic_factor_func(traffic_factor, traffic_start_time, cost)

optimal_route_traffic = simmulated_annealing_tsp(nodes_random, traffic_factor_with_traffic, traffic_polygon=traffic_jam_area, traffic_start_time=traffic_start_time, plot_cost=True)
```

```python
plot_tsp_route(nodes_random, optimal_route_traffic, traffic_polygon=traffic_jam_area)
```

```python
big_N = 5000
x_big = np.random.rand(big_N)
y_big = np.random.rand(big_N)

nodes_big_random = np.column_stack([x_big, y_big])

traffic_jam_area_big = np.array([[0.1, 0.9], [0.9, 0.1], [0.9, 0.9]]) # rectangle
traffic_factor_big = 100.0
traffic_start_time_big = 100.0

optimal_route_big_traffic = simmulated_annealing_tsp(nodes_big_random, traffic_polygon=traffic_jam_area_big, traffic_factor=traffic_factor_big, traffic_start_time=traffic_start_time_big, plot_cost=True)
```

```python
plot_tsp_route(nodes_big_random, optimal_route_big_traffic, traffic_polygon=traffic_jam_area_big)
```
